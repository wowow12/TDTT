Xây dựng lớp Shape với phương thức area().

Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().

Lấy π = 3.14.

Input

Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)

Dòng 2: tham số tương ứng.

Output

In ra diện tích với 2 chữ số thập phân.

```python
import math

class Shape:
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
        
    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r
        
    def area(self):
        return 3.14 * (self.r ** 2)

shape_type = input().strip()

if shape_type == "RECTANGLE":
    w, h = map(float, input().split())
    shape = Rectangle(w, h)
    print(f"{shape.area():.2f}")
elif shape_type == "CIRCLE":
    r = float(input())
    shape = Circle(r)
    print(f"{shape.area():.2f}")
```

Xây dựng lớp Vehicle(brand, speed).

Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle.

Viết phương thức info() cho mỗi lớp:

- Vehicle: brand-speed

- Car: brand-speed-seats

Input

Dòng 1: loại đối tượng (Vehicle hoặc Car)

Dòng 2: dữ liệu tương ứng.

Output

In ra chuỗi info().   



```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def info(self):
        return f"{self.brand}-{self.speed}"

class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats

    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"

obj_type = input().strip()
data = input().split()

if obj_type == "Vehicle":
    brand = data[0]
    speed = data[1]
    obj = Vehicle(brand, speed)
    print(obj.info())
elif obj_type == "Car":
    brand = data[0]
    speed = data[1]
    seats = data[2]
    obj = Car(brand, speed, seats)
    print(obj.info())
```

Xây dựng lớp Polygon với phương thức perimeter().

Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c.

Input

Một dòng chứa 3 số nguyên a b c.

Output

In ra chu vi tam giác.   

```python
class Polygon:
    def perimeter(self):
        pass

class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def perimeter(self):
        return self.a + self.b + self.c

a, b, c = map(int, input().split())
triangle = Triangle(a, b, c)
print(triangle.perimeter())
```

Xây dựng lớp Person(name, age) và lớp Student(name, age, school) kế thừa từ Person.

Phương thức describe():

- Person: name age

- Student: name age school

Input

Dòng 1: Person hoặc Student

Dòng 2: dữ liệu tương ứng.

Output

In ra kết quả describe().   

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def describe(self):
        return f"{self.name} {self.age}"

class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school

    def describe(self):
        return f"{self.name} {self.age} {self.school}"

obj_type = input().strip()
data = input().split()

if obj_type == "Person":
    name = data[0]
    age = data[1]
    obj = Person(name, age)
    print(obj.describe())
elif obj_type == "Student":
    name = data[0]
    age = data[1]
    school = data[2]
    obj = Student(name, age, school)
    print(obj.describe())
```
