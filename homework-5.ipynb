# W5A3.

Viết hàm kiểm tra một số có phải là số nguyên tố hay không? 

```python
import math
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    limit = int(math.sqrt(n))
    for i in range(3, limit + 1, 2):
        if n % i == 0:
            return False
    return True
if __name__ == "__main__":
    num = int(input("Nhập một số nguyên: "))
    if is_prime(num):
        print(f"{num} là số nguyên tố.")
    else:
        print(f"{num} không phải là số nguyên tố.")
```

# W5A4.

VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên. 

```python
import math
def is_perfect_number(n: int) -> bool:
    if n <= 1:
        return False
    tong_uoc = 1
    can_bac_hai = int(math.sqrt(n))
    for i in range(2, can_bac_hai + 1):
        if n % i == 0:
            tong_uoc += i
            if i != n // i:
                tong_uoc += n // i
    return tong_uoc == n
if __name__ == "__main__":
    test_number = int(input("Nhập số cần kiểm tra: "))
    
    if is_perfect_number(test_number):
        print(f"{test_number} là số hoàn hảo.")
    else:
        print(f"{test_number} không phải là số hoàn hảo.")
```

# W5A10.

Hai từ được gọi là đẳng cấu nếu các kí tự trong một từ có thể được ánh xạ để được từ còn lại. Ánh
xạ một kí tự nghĩa là thay đổi toàn bộ các kí tự đó trong từ bởi một kí tự khác. Thứ tự của các kí tự
được giữ không đổi. Hai kí tự khác nhau không thể ánh xạ đến cùng một kí tự, nhưng một kí tự có
thể ánh xạ thành chính nó. Ví dụ, từ "abca" và "zbxz" là hai từ đẳng cấu vì ta có thể ánh xạ 'a' thành
'z', 'b' thành 'b' và 'c' thành 'x'.
Viết hàm trả về \(true\) nếu hai từ \(a\) và \(b\) là đẳng cấu, ngược lại \(false\). Bạn có thể giả sử hai
từ đầu vào có cùng độ dài. 

```python
def is_isomorphic(a, b):
    map_a_to_b = {}
    map_b_to_a = {}
    
    for i in range(len(a)):
        char_a = a[i]
        char_b = b[i]
        
        if char_a in map_a_to_b:
            if map_a_to_b[char_a] != char_b:
                return False
        else:
            map_a_to_b[char_a] = char_b
            
        if char_b in map_b_to_a:
            if map_b_to_a[char_b] != char_a:
                return False
        else:
            map_b_to_a[char_b] = char_a
            
    return True
a,b=input().split()
print (is_isomorphic(a, b))
```

# W5A19.

Viết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.

```python
def is_magic_number(x):
    if x <= 0:
        return False
    
    while x >= 10:
        total = 0
        while x > 0:
            total += x % 10
            x //= 10
        x = total
        
    return x == 1
x = int(input())
print (is_magic_number(x))
```

# W5A25.

Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p):
    px, py = p
    signs = []
    
    for i in range(4):
        x1, y1 = quad[i]
        x2, y2 = quad[(i + 1) % 4]
        
        cross_product = (x2 - x1) * (py - y1) - (y2 - y1) * (px - x1)
        signs.append(cross_product)
        
    all_non_negative = True
    all_non_positive = True
    
    for s in signs:
        if s < 0:
            all_non_negative = False
        if s > 0:
            all_non_positive = False
            
    return all_non_negative or all_non_positive
vertices = []
for i in range(4):
    x, y = map(float, input().split())
    vertices.append((x, y))

px, py = map(float, input().split())
target_p = (px, py)

print(point_in_quadrilateral(vertices, target_p))
```

# W5A28.

Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
import math

def calculate_angle_abc(x1, y1, x2, y2, x3, y3):
    ux, uy = x1 - x2, y1 - y2
    vx, vy = x3 - x2, y3 - y2
    
    dot_product = ux * vx + uy * vy
    mag_u = math.sqrt(ux**2 + uy**2)
    mag_v = math.sqrt(vx**2 + vy**2)
    
    if mag_u == 0 or mag_v == 0:
        return 0.0
        
    cos_theta = dot_product / (mag_u * mag_v)
    cos_theta = max(-1.0, min(1.0, cos_theta))
    
    angle_rad = math.acos(cos_theta)
    return math.degrees(angle_rad)
x1, y1, x2, y2, x3, y3 = map(float,input().split())
print (calculate_angle_abc(x1, y1, x2, y2, x3, y3))
```

# W5A30.

Viết hàm tính hệ số nhị thức C(n, k)

```python
def nCr(n, k):
    if k < 0 or k > n:
        return 0
    if k == 0 or k == n:
        return 1
    if k > n // 2:
        k = n - k
        
    res = 1
    for i in range(k):
        res = res * (n - i) // (i + 1)
    return res
n,k = map(int,input().split())
print (nCr(n, k))
```

# W5A33.

Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, giá km, phí chờ, phụ phí ban đêm và giảm giá.

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        opening_fare = 11000
        km_rate_1 = 14500 
        km_rate_2 = 12500
    else:
        opening_fare = 13000
        km_rate_1 = 16000
        km_rate_2 = 14500 

    if km <= 0.6:
        fare = opening_fare
    elif km <= 25:
        fare = opening_fare + (km - 0.6) * km_rate_1
    else:
        fare = opening_fare + (25 - 0.6) * km_rate_1 + (km - 25) * km_rate_2

    if wait_minutes > 5:
        fare += (wait_minutes - 5) * 1000

    if start_hour >= 22 or start_hour < 5:
        fare *= 1.15

    if km > 50:
        fare *= 0.9

    return round(fare)
car_type, start_hour, km, wait_minutes = map(int,input().split())
print (smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

# W5A34.

Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n):
    if n <= 0:
        return 0
    total = 0
    for x in range(1, n + 1):
        s = str(x)
        if len(set(s)) == len(s):
            total += x
    return total
n = int(input())
print (sum_1_to_n(n))
```
