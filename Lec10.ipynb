```python
class Point3:
    def __init__(self,x_val,y_val,z_val):
        self.x = x_val
        self.y = y_val
        self.z = z_val
    def prt(self):
        print ("({} - {} - {})".format(self.x,self.y,self.z))
    def scale(self, scale_val):
        self.x += scale_val
        self.y += scale_val
        self.z += scale_val
    def distance(self,other_p): #Euclidean Distance
        dd = (self.x - other_p.x)**2 + (self.y - other_p.y)**2 + (self.z - other_p.z)**2
        d = dd**(1/2)
    def __str__(self):
        return "({} + {} + {})".format(self.x,self.y,self.z)
p1 = Point3(1,2,3) # Call dunder method __init__
p2 = Point3(4,5,6)
def printPoint(p):
    print ("({}, {}, {})".format(p.x,p.y,p.z))
p1.scale(2) # call Point3.scale(p1(self),2)
print (p1) # Call dunder method __str__

print (p2)
printPoint(p1)
p1.prt() # Call normal func/method prt()
p2.prt()
# Dunder python will auto-call it
# Normal need to call func
p1.distance(p2) # in ra nhu nao?
```

```python
class Counter:
    def __init__(self):
        self.count = 0
    def currCount(self):
        return self.count
    def incr(self):
        self.count+=1
    def reset (self):
        self.count=0
    def double2(count): # Static method
        return count*2

c = Counter()
print ("Current count: ", c.currCount())
c.incr() # Counter.incr(c(self))
c.incr()
print ("Current count:", c.currCount())
c.reset()
print ("Current count: ", c.currCount())
# c.double2() # -> Counter.double2(c(self))
Counter.double2(4)
```
